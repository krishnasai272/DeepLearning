```python
from google.colab import files
from PIL import Image
import numpy as np

# --- STEP 1: Upload the image from your computer ---
print("Please upload your image file:")
uploaded = files.upload()

# Automatically get the uploaded file name
image_path = list(uploaded.keys())[0]
original_img = Image.open(image_path)

print(f"\nSuccessfully loaded: {image_path}")
print("Original Image Mode:", original_img.mode)


# --- STEP 2: Convert Image to Pixels (NumPy Array) ---
pixel_array = np.array(original_img)

print("Pixel Array Shape (Height, Width, Channels):", pixel_array.shape)
print("First 2x2 pixels:\n", pixel_array[:2, :2])


# --- STEP 3: Convert Pixels back to Image ---
# Ensure data type is uint8 (standard for images)
reconstructed_img = Image.fromarray(pixel_array.astype('uint8'))


# --- STEP 4: Display and Save the Reconstructed Image ---
print("\nDisplaying Reconstructed Image:")
display(reconstructed_img)

# Save the file in the Colab environment
output_filename = "reconstructed_image.png"
reconstructed_img.save(output_filename)
print(f"Image saved as '{output_filename}' inside Colab.")

# Uncomment the line below if you want to automatically download it to your PC:
# files.download(output_filename)
```
