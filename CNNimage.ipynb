```python
# ==========================================
# PEN vs PENCIL IMAGE COMPARISON
# Run directly in Google Colab
# ==========================================

from google.colab import files
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np

print("Upload two images (Pen and Pencil)")
uploaded = files.upload()

# Get uploaded image names
image_names = list(uploaded.keys())

if len(image_names) != 2:
    raise Exception("Please upload exactly TWO images.")

img1 = Image.open(image_names[0]).convert("RGB")
img2 = Image.open(image_names[1]).convert("RGB")

# Resize both images
size = (224,224)
img1 = img1.resize(size)
img2 = img2.resize(size)

# Convert to numpy arrays
arr1 = np.array(img1).astype(np.float32)
arr2 = np.array(img2).astype(np.float32)

# Calculate similarity
difference = np.mean(np.abs(arr1 - arr2))
similarity = 100 - (difference / 255 * 100)

# Display images
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(img1)
plt.title("Image 1")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(img2)
plt.title("Image 2")
plt.axis("off")
plt.show()
print("="*50)
print("IMAGE COMPARISON RESULT")
print("="*50)
print(f"Similarity Score : {similarity:.2f}%")
if similarity > 85:
    print("Result : SAME OBJECT")
else:
    print("Result : DIFFERENT OBJECT")
```
