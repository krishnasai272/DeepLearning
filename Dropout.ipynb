```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models

# 1. Define your exact dataset from the image
study_hours = np.array([[1], [2], [3], [4], [5]], dtype=float)
marks = np.array([45, 55, 65, 75, 25], dtype=float)

# 2. Build a Neural Network model with a Dropout Layer
model = models.Sequential([
    # Input layer for 1 feature (Study Hours)
    layers.Input(shape=(1,)),

    # Hidden layer with 64 neurons and ReLU activation
    layers.Dense(64, activation='relu'),

    # --- DROPOUT LAYER ---
    # Randomly drops 20% of the neurons during each training batch step
    layers.Dropout(0.2),

    # Second hidden layer with 32 neurons
    layers.Dense(32, activation='relu'),

    # Output layer (predicting Marks)
    layers.Dense(1)
])

# 3. Compile the model
model.compile(
    optimizer='adam',
    loss='mean_squared_error' # Using MSE since we are predicting continuous numbers (marks)
)

# 4. Train the model (Dropout is active here)
print("Training Neural Network with Dropout...")
history = model.fit(
    study_hours,
    marks,
    epochs=500,  # Higher epochs to let the network learn despite dropout
    verbose=0    # Set to 1 if you want to see epoch-by-epoch loss
)

print("Training complete!")

# 5. Make a prediction for a new value (e.g., studying for 3.5 hours)
test_hours = np.array([[3.5]])
predicted_marks = model.predict(test_hours)

print(f"\nPredicted Marks for 3.5 hours of study: {predicted_marks[0][0]:.2f}")
```
