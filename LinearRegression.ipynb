```python
#LINEAR REGRESSION
import pandas as pd
from sklearn.linear_model import LinearRegression
#Dataset
data={
    'Time':[1,2,3,4,5,6],
    'Battery Charge':[12,24,36,48,60,71]
}
df=pd.DataFrame(data)
X=df[['Time']]
Y=df['Battery Charge']
#Create Model
model=LinearRegression()
model.fit(X,Y)
prediction=model.predict([[0]])
print(prediction)
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
#Dataset
data={
    'Hours':[1,2,3,4,5,6],
    'Marks':[35,45,55,65,75,85]
}
df=pd.DataFrame(data)
X=df[['Hours']]
Y=df['Marks']
#Create Model
model=LinearRegression()
model.fit(X,Y)
prediction=model.predict([[7]])
print(prediction)
```
