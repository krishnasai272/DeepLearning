```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

# 1. Define your data
study_hours = np.array([[1], [2], [3], [4], [5]])
marks = np.array([45, 55, 65, 75, 25])

# 2. Transform features to a high degree (Degree 4 polynomial for 5 data points)
# This gives the model enough "brain power/complexity" to completely memorize the data
poly = PolynomialFeatures(degree=4)
X_poly = poly.fit_transform(study_hours)

# 3. Train the model on the transformed data
model = LinearRegression()
model.fit(X_poly, marks)

# 4. Predict values for the training points to show 0 error (Memorization)
predictions = model.predict(X_poly)

print("Actual Marks:   ", marks)
print("Predicted Marks:", np.round(predictions, 2))
print("Result: The model has completely memorized every single point, including the noise!")

# 5. Optional: Plotting the overfitted curve
X_range = np.linspace(1, 5, 100).reshape(-1, 1)
X_range_poly = poly.transform(X_range)
y_range_pred = model.predict(X_range_poly)

plt.scatter(study_hours, marks, color='red', label='Training Data')
plt.plot(X_range, y_range_pred, color='blue', label='Overfitted Curve (Degree 4)')
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.legend()
plt.title('Overfitting Example: Memorizing the Noise')
plt.show()
```
